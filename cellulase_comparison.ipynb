# The general issue

-  Novel genome sequences emerge on a daily basis, but they receive little attention. 
-  Most genome assemblies are never cited (see figure below), and about 4/5 of eukaryotic genome assemblies available at NCBI Genbank are not annotated.
<figure>
<p align="left">
<img src="docs/figures/new_species_annotated_plot.jpg" alt="Growth of genomic resources" width=600/>
<figcaption><b>Figure 2.</b> Growth of genomic resources over time.</figcaption>
</p>
</figure>

-  Orthology based phylogenetic profiles based on public orthology databases provide a rapid overview of the taxonomic distribution and the evolutionary age of a protein.
-  the substantial computational overhead of computing all-against-all orthology assignments limits taxon sampling a fraction of the currently available genome assemblies. 
   -  for evolutionarily old and well conserved proteins, this is negiligible
   -  for evolutionarily younger proteins, or for such with a patchy presence-absence pattern, e.g., due to an HGT history, the limited taxon sampling in the public databases becomes an issue. 
- the following example highlights the issue

## Example
We use the fungal GH16 family glycoside hydrolase [QRW19682](https://www.ncbi.nlm.nih.gov/protein/QRW19682) encoded in the genome of [Rhizoctonia solani](https://www.ncbi.nlm.nih.gov/Taxonomy/Browser/wwwtax.cgi?command=show&mode=node&id=456999&lvl=) as an example. According to the OMA browser, orthologs of this protein are largly confined to fungi (See OMA HOG [HOG:0855835](https://omabrowser.org/oma/hog/HOG:F0855835/Eukaryota/iham/)). There is sporadic evidence for orthologs in Amboebozoa and in the SAR group.

None of the currently 365 animals species (OMA release 06/2026) harbour an ortholog to the fungal GH16 glycosyl hydrolase. **It is tempting to assume that this observation generalizes to all animals**.
