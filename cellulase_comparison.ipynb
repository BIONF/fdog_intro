# The general issue

Novel genome sequences emerge on a daily basis, but they receive little attention.
<p align="left">
  <img src="docs/figures/new_species_annotated.png " alt="Growth of genomic resources"/>
</p>
Orthology based phylogenetic profiles based on public orthology databases provide a great overview for the presence-absence pattern of proteins. However, due to the substantial computational overhead of computing all-against-all orthology assignments, taxon sampling is limited to only a fraction of the currently available genome assemblies. For evolutionarily old and well conserved proteins, this is not so much of an issue. For evolutionarily young proteins, or such with a patchy presence-absence pattern, e.g., due to an HGT history, the limited taxon sampling in the public databases becomes an issue. 

## Example
The phylogenetic profile of the GH16 family glycosilase QRW19682 of [Rhizoctonia solani](https://www.ncbi.nlm.nih.gov/Taxonomy/Browser/wwwtax.cgi?command=show&mode=node&id=456999&lvl=) suggests that this protein is mainly confined to fungi (See OMA HOG [HOG:0855835](https://omabrowser.org/oma/hog/HOG:F0855835/Eukaryota/iham/)). There is sporadic evidence for orthologs in Amboebozoa and in the SAR group, but not in animals.
