# The general issue

-  Novel genome sequences emerge on a daily basis, but they receive little attention. 
-  Most genome assemblies are never cited (see figure below), and about 4/5 of eukaryotic genome assemblies available at NCBI Genbank are not annotated.
<figure>
<p align="left">
<img src="docs/figures/new_species_annotated_plot.jpg" alt="Growth of genomic resources" width=600/>

</p>
</figure>

-  Orthology based phylogenetic profiles based on public orthology databases provide a rapid overview of the taxonomic distribution and the evolutionary age of a protein
-  The substantial computational overhead of all-against-all orthology assignments limits taxon sampling to a fraction of the currently available genome assemblies 
   -  for evolutionarily old and well conserved proteins, this is negiligible
   -  for evolutionarily younger proteins, or for such with a patchy presence-absence pattern, e.g., due to an HGT history, the limited taxon sampling in the public databases becomes an issue. 
- the following example highlights the issue

The relevant directories are:
  - _coreTaxa_dir_ -> contains the protein sets of the species that should be considered for compilation of the core orthologous groups (training data for the pHMMs; see Fig. 4 left tree)
  - _searchTaxa_dir_ -> contains the protein sets of the species for the final targeted ortholog search (see Fig. 4 right tree)
  - _annotation_dir_ -> contains the feature architectures for all proteins in the individual protein sets (see [Dosch et al. 2023](https://academic.oup.com/bioinformatics/article/doi/10.1093/bioinformatics/btad226/7135831))
  - _assembly_dir_ -> contains the unannotated genome assemblies used by fDOG-Assembly (Muelbaier et al. 2026 NAR accepted)

- _infile.fa_ -> seed sequence for the fDOG search in fasta format
- _jobname_ -> name of the job. This will be the base name of all output files
- _refspec_ -> Reference species for the fDOG search. This should be the species the seed sequence was derived from. **Watch out: This is not explicitely tested**
- _outpath_ -> path to the output directory. fDOG will create it if it does not exist.

## Package 2: Use fDOG to determine the phylogenetic profile of two plant cell wall enzymes from the fungus _Rhizoctonia solani_ 
### Background
Rhizoctonia solani (NCBI Taxonomy ID: [456999](https://www.ncbi.nlm.nih.gov/Taxonomy/Browser/wwwtax.cgi?command=show&mode=node&id=456999&lvl=)) has substantial plant cell wall–degrading capabilities that contribute to its ability to infect and colonize plant tissues. It produces carbohydrate-active enzymes (CAZymes), including pectinases, cellulases, hemicellulases, and other enzymes that break down components of the plant cell wall. These enzymes facilitate penetration and tissue maceration while releasing nutrients that can support fungal growth. The repertoire and expression of these enzymes can vary among isolates and host plants, contributing to the pathogen’s broad host range and virulence.

You should now be familiar with most of th details of the fDOG call. What is new?

## Package 3: Targeted ortholog search in unannotated genome assemblies
While fDOG searches for orthologs in protein collections, fDOG-assembly (fDA) can identify orthologs directly in genome assemblies. We will use this tool to assess whether analysing a further species from the genus _Bradysia_ supports the findings from the ortholog search in _B. coprophila_.

### parameterizing the fDOG-assembly search
Most of the settings in the command call you already know. What is new?
- _coregroupPath_ -> Other than fDOG, fDA cannot compile core orthologous groups on the fly. Instead, it uses an existing core ortholog group. This option allows to specify this path. **NOTE**: fDA expects a directory named exactly as the string provided with the option _gene_ 
- _assemblyPath_ -> path to the directory where the genome assembly is located that you want to search in
- _fast_ -> fDA has several search modes. In the fast mode, miniprot is used to identify candidate orthologs in the genome assembly, and no further gene prediction, e.g. via Augustus, is performed.

```python
# Run fDOG Assembly
!fdog.assembly --gene QRW19256 --refSpec RHISO@456999@016906535_1 --coregroupPath pcd_output/core_orthologs/ --assemblyPath assembly_dir/ --dataPath ./ --out pcd_output/ --fast
!fdog.assembly --gene QRW19682 --refSpec RHISO@456999@016906535_1 --coregroupPath pcd_output/core_orthologs/ --assemblyPath assembly_dir/ --dataPath ./ --out pcd_output/ --fast
```
